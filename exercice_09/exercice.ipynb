```python
import pandas as pd

titanic = pd.read_csv("../exercice_08/data/titanic_propre.csv")

titanic.head()
```

```python
titanic.groupby(["class", "sex"])["survived"].mean().mul(100).round(2)
```

```python
titanic.pivot_table(
    index="class",
    columns="sex",
    values="survived",
    aggfunc="mean"
).mul(100).round(2)
```

les femmes ont mieux survecu que les hommes dans les trois classes (first , second , third ) du Titanic

```python
titanic["tranche_age"] = pd.cut(
    titanic["age"],
    bins=[0, 18, 60, float("inf")],
    labels=["Enfants", "Adultes", "60 ans et plus"],
    right=False
)

titanic.groupby("tranche_age", observed=False)["survived"].mean().mul(100).round(2)
```

les enfants ont un taux de survie plus eleve que les adulte et les 60 ans et plus 

```python
titanic.groupby("embark_town")["fare"].mean().sort_values(ascending=False).round(2)
```

```python
pd.crosstab(
    titanic["embark_town"],
    titanic["class"],
    normalize="index"
).mul(100).round(2)
```

le port Cherbourg presente le prix moyen des billets le plus eleves 

```python
titanic.groupby("alone")["survived"].mean().mul(100).round(2)
```

```python
titanic.groupby("alone")["survived"].mean().mul(100).round(2).rename(
    index={True: "Voyage seul", False: "Voyage accompagné"}
)
```

Les passagers voyageant accompagnés ont un taux de survie plus élevé que les autres.
