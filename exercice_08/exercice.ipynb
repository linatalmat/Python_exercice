```python
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset("titanic")
```

```python
titanic.info()
```

```python
titanic["age"] = titanic.groupby(["class","sex"])["age"].transform(lambda x: x.fillna(x.median()))
```

```python
titanic["embark_town"] = titanic["embark_town"].fillna(titanic["embark_town"].mode()[0])
```

```python
titanic = titanic.drop(columns=["deck"])
```

```python
titanic.isnull().sum()
```

```python
print("nombre des doublons : ",titanic.duplicated().sum())
```

```python
titanic = titanic.drop_duplicates()
```

```python
titanic.duplicated().sum()
```

```python

titanic = titanic.drop(columns=["embarked"])


titanic["survived"] = titanic["survived"].astype("int64")
titanic["pclass"] = titanic["pclass"].astype("int64")
```

```python
print(titanic["fare"].describe())

# Repérer les billets dont le prix dépasse la borne supérieure de l'IQR
Q1 = titanic["fare"].quantile(0.25)
Q3 = titanic["fare"].quantile(0.75)
IQR = Q3 - Q1

borne_sup = Q3 + 1.5 * IQR

billets_extremes = titanic[titanic["fare"] > borne_sup]
print(billets_extremes[["fare", "class"]])
```
