```python
 
titanic.groupby("sex")["survived"].mean() * 100
```

```python

femmes = titanic[(titanic["sex"] == "female") & (titanic["class"] == "First")]
print("Femmes de 1re classe :", femmes["survived"].mean() * 100, "%")


hommes = titanic[(titanic["sex"] == "male") & (titanic["class"] == "Third")]
print("Hommes de 3e classe :", hommes["survived"].mean() * 100, "%")
```
