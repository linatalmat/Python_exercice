```python
import pandas as pd

titanic_propre = pd.read_csv("../exercice_08/data/titanic_propre.csv")

ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"]
})
  
print(ports)
```

```python
print("Nombre de passagers avant la jointure :", len(titanic_propre))
```

```python
titanic_enrichi = titanic_propre.merge(
    ports,
    on="embark_town",
    how="left",
    validate="many_to_one"
)

print(titanic_enrichi.head())
```

```python
print("Avant :", len(titanic_propre))
print("Après :", len(titanic_enrichi))
```

```python
print(titanic_propre.columns.tolist())
print(titanic_enrichi.columns.tolist())
```

```python
print(titanic_enrichi["pays"].value_counts(dropna=False))
```

```python
taux_survie = titanic_enrichi.groupby("pays")["survived"].mean()

print(taux_survie)
```

```python
print((taux_survie * 100).round(2))
```

```python
ports_sans_queenstown = ports[
    ports["embark_town"] != "Queenstown"
]

print(ports_sans_queenstown)
```

```python
titanic_left = titanic_propre.merge(
    ports_sans_queenstown,
    on="embark_town",
    how="left"
)

print("Nombre de passagers avec left :", len(titanic_left))
print(titanic_left.head())
```

```python
titanic_inner = titanic_propre.merge(
    ports_sans_queenstown,
    on="embark_town",
    how="inner"
)

print("Nombre de passagers avec inner :", len(titanic_inner))
print(titanic_inner.head())
```

```python
print("Passagers au départ :", len(titanic_propre))
print("Après jointure left :", len(titanic_left))
print("Après jointure inner :", len(titanic_inner))
```

```python
print(titanic_left["pays"].isna().sum())
```
